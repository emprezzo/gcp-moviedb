**Actor**

```python
%%bigquery
create or replace table movie_csp.Actor(
  credit_id STRING not null,
  person_id INTEGER not null,
  movie_id INTEGER not null,
  name STRING,
  character STRING,
  gender INTEGER,
  credit_order INTEGER,
  data_source STRING not null,
  load_time TIMESTAMP not null,
  effective_time TIMESTAMP default current_timestamp() not null,
  discontinue_time TIMESTAMP,
  status_flag BOOL not null,
  primary key (credit_id, effective_time) not enforced)
  as select *, current_timestamp, null, true
  from movie_stg.Actor;
```

```python
%%bigquery
select c.column_name, c.data_type, c.is_nullable, c.column_default, k.constraint_name
from movie_csp.INFORMATION_SCHEMA.COLUMNS c left join movie_csp.INFORMATION_SCHEMA.KEY_COLUMN_USAGE k
on c.table_name = k.table_name and c.column_name = k.column_name
where c.table_name = 'Actor'
order by c.ordinal_position;
```

```python
%%bigquery
select * from movie_csp.Actor
limit 5
```
